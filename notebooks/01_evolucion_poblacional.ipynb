```python
import pandas as pd
import numpy as np

# Cargar los datos
df = pd.read_csv("../data/raw/annual-population-growth/annual-population-growth.csv")

# Ver las entidades únicas disponibles
print("Entidades disponibles:")
print(df["Entity"].unique())
```

## Actividad 2 — Exploración de datos

### Primeras filas del dataset

```python
# Primeras filas
print("Primeras filas:")
print(df.head())

# Dimensiones
print("\nDimensiones:")
print(df.shape)

# Nombres de columnas
print("\nNombres de columnas:")
print(df.columns.tolist())

# Tipos de datos
print("\nTipos de datos:")
print(df.dtypes)

# Estadísticas descriptivas
print("\nEstadísticas descriptivas:")
print(df.describe())

# Cantidad de valores faltantes
print("\nCantidad de valores faltantes:")
print(df.isnull().sum())

# Cantidad de entidades
print("\nCantidad de entidades:")
print(df["Entity"].nunique())
```

### Filtrado de datos específicos

```python
# Filtrar año 2020
df_2020 = df[df["Year"] == 2020][["Entity", "Year"]]
print("Datos para el año 2020:")
print(df_2020.head(3))

# Filtrar datos de México entre 2005 y 2020
df_mexico_2005_2020 = df[(df["Year"] >= 2005) & (df["Year"] <= 2020) & (df["Entity"] == "Mexico")]
print("\nDatos de México (2005-2020):")
print(df_mexico_2005_2020.shape)

# Extraer columnas específicas
df_ext = df_mexico_2005_2020[["Entity", "Year", "Annual change in population"]]
print("\nPrimeras filas de México:")
print(df_ext.head(3))

# Calcular media
print("\nMedia de cambio anual (México 2005-2020):")
print(df_ext.mean(numeric_only=True))
```

## Actividad 3 — Estadística descriptiva

### Propósito
Aplicar medidas estadísticas básicas a una variable poblacional.

### Instrucciones
Selecciona un año del periodo histórico y filtra los datos correspondientes. Calcula sobre la variable de cambio anual de población:
- Media
- Mediana
- Varianza
- Desviación estándar
- Mínimo
- Máximo

```python
# Seleccionar el año 2023 para análisis
df_2023 = df[df["Year"] == 2023]

# Calcular estadísticas
print("Estadísticas para el año 2023:")
print(f"Media: {df_2023['Annual change in population'].mean():.2e}")
print(f"Mediana: {df_2023['Annual change in population'].median():.2e}")
print(f"Desviación estándar: {df_2023['Annual change in population'].std():.2e}")
print(f"Varianza: {df_2023['Annual change in population'].var():.2e}")
print(f"Mínimo: {df_2023['Annual change in population'].min():.2e}")
print(f"Máximo: {df_2023['Annual change in population'].max():.2e}")
```

### Identificación de países extremos

```python
# Definir regiones para filtrar solo países
regiones = [
    "World",
    "Africa (UN)",
    "Americas (UN)",
    "Asia (UN)",
    "Europe (UN)",
    "Oceania (UN)",
    "Less developed regions, excluding China",
    "Less developed regions, excluding least developed countries",
    "Less developed regions",
    "Lower-middle-income countries",
    "Upper-middle-income countries",
    "Least developed countries"
]

# Filtrar solo países
df_paises = df[~df["Entity"].isin(regiones)]

# Encontrar país con mínimo y máximo cambio
minimo = df_paises.loc[df_paises["Annual change in population"].idxmin()]
maximo = df_paises.loc[df_paises["Annual change in population"].idxmax()]

print("País con el menor cambio anual:")
print(minimo)
print("\nPaís con el mayor cambio anual:")
print(maximo)
```

### Análisis de Venezuela (1983-2021)

```python
# Filtrar datos de Venezuela entre 1983 y 2021
df_venezuela = df[
    (df["Entity"] == "Venezuela") &
    (df["Year"] >= 1983) &
    (df["Year"] <= 2021)
].copy()

# Extraer columnas y eliminar duplicados
df_venezuela = df_venezuela[["Entity", "Year", "Annual change in population"]].drop_duplicates()

# Calcular estadísticos
estadisticos_venezuela = df_venezuela["Annual change in population"].agg([
    "count", "mean", "median", "var", "std", "min", "max"
])

print("Estadísticos para Venezuela (1983-2021):")
print(estadisticos_venezuela)
```
