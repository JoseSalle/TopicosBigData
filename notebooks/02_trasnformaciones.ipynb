```python
import pandas as pd
import requests
import numpy as np

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()

df.head()
```

```python
df.rename(columns={'entity': "Pais",  'code': "Código", 'year': "Año" ,'life_expectancy_0': "Esperanza de Vida"}, inplace=True)
```

```python
df.head(3)
```

```python
df.reset_index(inplace=True)
```

```python
df.reset_index(inplace=True)
df.set_index(["Pais", "Año"], inplace = True)
```

```python
df.loc[df.index.get_level_values('Pais').str.contains("U", case = False, regex = True)]
```

```python
df.loc[df["Código"].isnull()]
```

```python
df.dropna(subset=["Código"], inplace=True)
df.isnull().sum()
```

```python
resultado = (
    df.groupby("Pais")["Esperanza de Vida"].mean().sort_values(ascending=True)
)

resultado2020 = (
    df.loc[df.index.get_level_values("Año") == 2020]
    .groupby("Pais")["Esperanza de Vida"]
    .mean()
    .sort_values(ascending=True)
)

resultado2020.head(10)
```
